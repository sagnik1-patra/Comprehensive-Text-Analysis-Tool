```python
from pathlib import Path
from collections import Counter, defaultdict
import re, json, math, csv, sys
from urllib.parse import urlparse
import numpy as np
import pandas as pd
import matplotlib
# Notebook: use the existing inline Matplotlib backend (do not force TkAgg).
import matplotlib.pyplot as plt
from matplotlib.widgets import RadioButtons

# Jupyter has no __file__. Detect the folder containing the four BBC files.
WINDOWS_FOLDER = Path(r"C:\Users\sagni\Downloads\Comprehensive Text Analysis Tool")
CANDIDATES = [WINDOWS_FOLDER, Path.cwd(), Path.cwd() / "Comprehensive Text Analysis Tool", Path('/mnt/data')]
ROOT = next((p for p in CANDIDATES if all((p / f).is_file() for f in ('bbc.mtx','bbc.terms','bbc.classes','bbc.docs'))), None)
if ROOT is None:
    raise FileNotFoundError("BBC input files not found. Set ROOT = Path(r'C:\\Users\\sagni\\Downloads\\Comprehensive Text Analysis Tool') to your actual folder.")
print('Dataset folder:', ROOT)
OUT = ROOT / 'results'
OUT.mkdir(exist_ok=True)
plt.rcParams.update({'figure.dpi':110,'savefig.dpi':120,'font.size':10})
COLORS = ['#2463a5','#e5a43c','#34958d','#9c64a6','#d86c68']
STOP = set('a an and are as at be been but by for from had has have he her his i in is it its of on or our she that the their them there these they this to was were will with you your not can do did more than into over after before about who what when where which also one two three new said say says would could should'.split())
POSITIVE=set('good great growth gain gains success successful win wins best improve improved benefit strong positive profit profits rise rising'.split())
NEGATIVE=set('bad loss losses decline crisis fail failed failure weak negative fall falling risk risks drop dropped'.split())

def read_lines(path):
    """Read UTF-8/ASCII text with fallback for imperfect encodings."""
    for enc in ('utf-8-sig','ascii','cp1252','latin-1'):
        try: return path.read_text(encoding=enc).splitlines()
        except UnicodeError: pass
    return path.read_text(encoding='utf-8',errors='replace').splitlines()

def load_dataset():
    """Read MatrixMarket coordinate matrix without scipy; rows=terms, columns=documents."""
    terms=[x.strip() for x in read_lines(ROOT/'bbc.terms')]
    docs=[x.strip() for x in read_lines(ROOT/'bbc.docs')]
    classes={}
    category_names=[]
    for line in read_lines(ROOT/'bbc.classes'):
        if line.startswith('%Clusters'):
            category_names=line.split(' ',2)[-1].split(',')
        elif line and not line.startswith('%'):
            bits=line.split()
            if len(bits)==2 and all(s.isdigit() for s in bits):
                classes[int(bits[0])]=int(bits[1])
    matrix_lines=iter(read_lines(ROOT/'bbc.mtx'))
    header=next(matrix_lines)
    if not header.startswith('%%MatrixMarket matrix coordinate'): raise ValueError('Expected MatrixMarket coordinate format')
    dims=next(x for x in matrix_lines if x.strip() and not x.startswith('%'))
    nterms,ndocs,nnz=map(int,dims.split())
    if nterms!=len(terms) or ndocs!=len(docs): raise ValueError('Matrix dimensions do not match vocabulary/document list')
    doc_counts=[Counter() for _ in docs]
    for line in matrix_lines:
        if not line.strip() or line.startswith('%'): continue
        i,j,v=line.split()[:3]
        count=float(v)
        if count<0: raise ValueError('Negative count')
        doc_counts[int(j)-1][terms[int(i)-1]]+=count
    return terms,docs,classes,category_names,doc_counts,nnz

def tokenize(text):
    return re.findall(r"[A-Za-z]+(?:'[A-Za-z]+)?|\d+(?:\.\d+)?",text.lower())

def extract_patterns(text):
    """Extract patterns from original text only (not synthetic bag-of-words)."""
    emails=re.findall(r'(?<![\w.+-])[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}(?!\w)',text)
    urls=re.findall(r'https?://[^\s<>"\']+|www\.[^\s<>"\']+',text,re.I)
    urls=[u.rstrip('.,;:!?)') for u in urls]
    phones=re.findall(r'(?<!\w)(?:\+1[-.\s]?)?(?:\(\d{3}\)|\d{3})[-.\s]?\d{3}[-.\s]?\d{4}(?!\d)',text)
    normalized=[]
    for phone in phones:
        d=re.sub(r'\D','',phone)
        if len(d)==11 and d[0]=='1': d=d[1:]
        if len(d)==10: normalized.append(f'({d[:3]}) {d[3:6]}-{d[6:]}')
    dates=re.findall(r'\b(?:\d{1,2}/\d{1,2}/\d{4}|\d{1,2}-\d{1,2}-\d{4}|(?:January|February|March|April|May|June|July|August|September|October|November|December)\s+\d{1,2},?\s+\d{4})\b',text,re.I)
    money=re.findall(r'(?<!\w)[$£€]\s?\d[\d,]*(?:\.\d{1,2})?',text)
    domains=Counter(urlparse(u if u.startswith('http') else 'https://'+u).netloc.lower() for u in urls)
    return dict(email_count=len(emails),emails=sorted(set(emails)),phone_count=len(phones),phones=sorted(set(normalized)),hashtags=sorted(set(re.findall(r'(?<!\w)#[A-Za-z][\w]*',text))),mentions=sorted(set(re.findall(r'(?<!\w)@[A-Za-z][\w]*',text))),urls=sorted(set(urls)),domains=dict(domains),dates=sorted(set(dates)),currencies=sorted(set(money)))

def savefig(name):
    plt.tight_layout();plt.savefig(OUT/name,bbox_inches='tight');
    plt.show()  # Display every saved plot in the notebook
    plt.close()

def barplot(labels,values,title,file,horizontal=False):
    plt.figure(figsize=(11,6))
    if horizontal:
        y=np.arange(len(labels));plt.barh(y,values,color=COLORS[0]);plt.yticks(y,labels);plt.gca().invert_yaxis()
        for j,v in enumerate(values):plt.text(v,j,f' {v:,.0f}',va='center',fontsize=8)
    else:
        x=np.arange(len(labels));plt.bar(x,values,color=COLORS[0]);plt.xticks(x,labels,rotation=65,ha='right')
        for j,v in enumerate(values):plt.text(j,v,f'{v:,.0f}',ha='center',va='bottom',fontsize=7)
    plt.title(title);plt.ylabel('Frequency' if not horizontal else '');plt.xlabel('Frequency' if horizontal else '')
    savefig(file)

def ngrams(tokens,n): return Counter(tuple(tokens[i:i+n]) for i in range(len(tokens)-n+1))

def analyze():
    terms,docs,classes,names,counts,nnz=load_dataset()
    total=Counter()
    metadata=[];categories=defaultdict(Counter)
    for i,(doc,c) in enumerate(zip(docs,counts)):
        cat=names[classes[i]] if i in classes and classes[i]<len(names) else 'unknown'
        total.update(c);categories[cat].update(c)
        wc=int(sum(c.values()));chars=int(sum(len(w)*f for w,f in c.items()))
        metadata.append(dict(document_id=i+1,filename=doc,category=cat,word_count=wc,character_count=chars,unique_words=len(c),average_word_length=round(chars/wc,3) if wc else 0,vocabulary_richness=round(len(c)/wc,5) if wc else 0,sentence_count=np.nan,average_sentence_length=np.nan))
    df=pd.DataFrame(metadata);df.to_csv(OUT/'document_statistics.csv',index=False)
    freq=pd.DataFrame([{'word':w,'frequency':int(f),'length':len(w),'first_char':w[0] if w else '', 'vowel_count':sum(ch in 'aeiou' for ch in w)} for w,f in total.items()]).set_index('word').sort_values('frequency',ascending=False)
    freq.to_csv(OUT/'word_statistics.csv');freq.head(50).to_csv(OUT/'top_50_words.csv')
    freq[freq.length>10].to_csv(OUT/'words_longer_than_10.csv')
    freq.groupby('length').agg(word_types=('frequency','size'),total_occurrences=('frequency','sum'),mean_frequency=('frequency','mean')).to_csv(OUT/'groupby_word_length.csv')
    freq.groupby('first_char').agg(word_types=('frequency','size'),total_occurrences=('frequency','sum')).to_csv(OUT/'groupby_first_character.csv')
    pd.pivot_table(freq.reset_index(),index='length',columns='first_char',values='frequency',aggfunc='sum',fill_value=0).to_csv(OUT/'pivot_length_by_first_character.csv')
    pd.crosstab(freq.length,freq.vowel_count).to_csv(OUT/'crosstab_length_vowels.csv')
    df.groupby('category').agg(documents=('document_id','size'),total_words=('word_count','sum'),mean_words=('word_count','mean'),mean_richness=('vocabulary_richness','mean')).to_csv(OUT/'category_summary.csv')
    top=Counter({w:f for w,f in total.items() if w not in STOP and len(w)>2}).most_common(20)
    barplot([w for w,_ in top],[f for _,f in top],'Top 20 meaningful terms','01_top20_words.png')
    barplot([w for w,_ in top],[f for _,f in top],'Top 20 meaningful terms (horizontal)','02_top20_horizontal.png',True)
    lens=np.array([len(w) for w in total]);weights=np.array([f for f in total.values()]);mu=np.average(lens,weights=weights);sigma=np.sqrt(np.average((lens-mu)**2,weights=weights))
    plt.figure(figsize=(10,5));bins=np.arange(0,max(lens)+2)-.5;plt.hist(lens,bins=bins,weights=weights,density=True,color=COLORS[0],alpha=.75,label='Observed')
    x=np.linspace(0,max(lens),300)
    if sigma>0:plt.plot(x,np.exp(-.5*((x-mu)/sigma)**2)/(sigma*np.sqrt(2*np.pi)),color=COLORS[1],label='Normal reference')
    plt.title('Word length distribution (weighted by occurrences)');plt.xlabel('Characters per token');plt.ylabel('Density');plt.legend();savefig('03_word_length_histogram.png')
    chars=Counter();letters=Counter();punct=Counter()
    for w,f in total.items():
        for ch in w:
            chars[ch]+=f
            if ch.isalpha(): letters[ch]+=f
            elif not ch.isalnum():punct[ch]+=f
    topchars=chars.most_common(15)
    barplot([x for x,_ in topchars],[y for _,y in topchars],'Top 15 characters in vocabulary tokens','04_character_frequency.png')
    barplot([x for x,_ in letters.most_common(20)],[y for _,y in letters.most_common(20)],'Letter frequencies','05_letter_frequency.png')
    if punct:barplot([x for x,_ in punct.most_common(15)],[y for _,y in punct.most_common(15)],'Punctuation in stored terms','06_punctuation_frequency.png')
    else:
        plt.figure(figsize=(8,3));plt.text(.5,.5,'No punctuation retained in dataset vocabulary',ha='center',va='center');plt.axis('off');savefig('06_punctuation_frequency.png')
    plt.figure(figsize=(10,4));plt.boxplot(df.word_count,vert=False);plt.title('Document word count distribution (sentence lengths unavailable)');plt.xlabel('Words per document');savefig('07_document_length_boxplot.png')
    ranks=np.arange(1,len(freq)+1);values=freq.frequency.to_numpy()
    valid=values>0
    slope,intercept=np.polyfit(np.log10(ranks[valid]),np.log10(values[valid]),1)
    plt.figure(figsize=(9,5));plt.loglog(ranks,values,label='Observed term frequency',color=COLORS[0]);plt.loglog(ranks,10**intercept*ranks**slope,'--',label=f'Log-log fit: slope {slope:.2f}',color=COLORS[1]);plt.title('Zipf rank-frequency analysis');plt.xlabel('Rank');plt.ylabel('Frequency');plt.legend();savefig('08_zipf_law.png')
    plt.figure(figsize=(9,5));cats=df.groupby('category').word_count.mean().sort_values(ascending=False);plt.bar(cats.index,cats.values,color=COLORS[:len(cats)]);plt.title('Mean document length by category');plt.ylabel('Mean word count');savefig('09_category_comparison.png')
    plt.figure(figsize=(9,5));plt.scatter(df.word_count,df.vocabulary_richness,alpha=.25,s=12,color=COLORS[0]);plt.xlabel('Document word count');plt.ylabel('Type-token ratio');plt.title('Document length vs vocabulary richness');savefig('10_richness_scatter.png')
    fig,ax=plt.subplots(2,2,figsize=(13,9));ax[0,0].bar([x for x,_ in top[:10]],[y for _,y in top[:10]],color=COLORS[0]);ax[0,0].tick_params(axis='x',rotation=55);ax[0,0].set_title('Top terms')
    ax[0,1].hist(df.word_count,bins=35,color=COLORS[1]);ax[0,1].set_title('Document lengths')
    ax[1,0].bar(cats.index,cats.values,color=COLORS[2]);ax[1,0].tick_params(axis='x',rotation=30);ax[1,0].set_title('Mean words by category')
    ax[1,1].scatter(df.word_count,df.vocabulary_richness,s=8,alpha=.3,color=COLORS[3]);ax[1,1].set_title('Vocabulary richness')
    fig.suptitle('BBC News Text Analysis Dashboard',fontsize=17);savefig('11_dashboard_2x2.png')
    # Co-occurrence is NOT an ordered bigram; preserve that distinction.
    cooccur=Counter()
    for c in counts:
        key=[w for w,_ in c.most_common(12) if w not in STOP and len(w)>2][:8]
        for i in range(len(key)):
            for j in range(i+1,len(key)):cooccur[tuple(sorted((key[i],key[j])))]+=1
    pd.DataFrame([{'term_a':a,'term_b':b,'document_cooccurrence':v} for (a,b),v in cooccur.most_common(100)]).to_csv(OUT/'top_document_cooccurrences.csv',index=False)
    sentiment=[]
    for doc,c in zip(docs,counts):
        pos=sum(c.get(w,0) for w in POSITIVE);neg=sum(c.get(w,0) for w in NEGATIVE)
        sentiment.append({'document':doc,'positive_keyword_count':int(pos),'negative_keyword_count':int(neg),'keyword_score':int(pos-neg)})
    pd.DataFrame(sentiment).to_csv(OUT/'sentiment_keyword_scores.csv',index=False)
    # No original text is provided; use optional original .txt files for genuine regex, sentence, n-gram results.
    originals=ROOT/'original_articles';originals.mkdir(exist_ok=True)
    textfiles=sorted(originals.glob('*.txt'))
    patterns={};sent_lens=[];all_bigrams=Counter();all_trigrams=Counter();all_clean=[]
    for f in textfiles:
        raw='\n'.join(read_lines(f));patterns[f.name]=extract_patterns(raw)
        sentences=[s.strip() for s in re.split(r'(?<=[.!?])\s+',raw) if s.strip()]
        sent_lens.extend(len(tokenize(s)) for s in sentences)
        tokens=tokenize(raw);all_clean.append(' '.join(tokens));all_bigrams.update(ngrams(tokens,2));all_trigrams.update(ngrams(tokens,3))
    (OUT/'cleaned_processed_text.txt').write_text('\n'.join(all_clean) if all_clean else 'Original ordered text unavailable: BBC .mtx contains term frequencies, not full articles. See word_statistics.csv for authentic processed tokens.\n',encoding='utf-8')
    (OUT/'extracted_patterns.json').write_text(json.dumps({'source':'Original .txt files in original_articles/ only','files':patterns,'note':'Regex results are not inferred from bag-of-words data.'},indent=2),encoding='utf-8')
    for label,data in [('bigrams',all_bigrams),('trigrams',all_trigrams)]:
        pd.DataFrame([{'ngram':' '.join(k),'frequency':v} for k,v in data.most_common(100)],columns=['ngram','frequency']).to_csv(OUT/f'{label}.csv',index=False)
    plt.figure(figsize=(10,4))
    if sent_lens:plt.boxplot(sent_lens,vert=False);plt.title('Sentence lengths from original articles');plt.xlabel('Words')
    else:plt.text(.5,.5,'Original sentences unavailable in bag-of-words dataset',ha='center',va='center');plt.axis('off')
    savefig('12_sentence_length_boxplot.png')
    plt.figure(figsize=(10,4))
    if textfiles:
        labels=[k for k,_ in all_bigrams.most_common(10)];vals=[v for _,v in all_bigrams.most_common(10)]
        plt.barh([' '.join(x) for x in labels][::-1],vals[::-1],color=COLORS[0]);plt.title('Ordered bigrams from original articles')
    else:plt.text(.5,.5,'Ordered bigrams/trigrams require original article text',ha='center',va='center');plt.axis('off')
    savefig('13_bigram_analysis.png')
    vowels=sum(letters.get(c,0) for c in 'aeiou');consonants=sum(v for c,v in letters.items() if c.isascii() and c.isalpha() and c not in 'aeiou')
    outliers=df[df.word_count>df.word_count.quantile(.75)+1.5*(df.word_count.quantile(.75)-df.word_count.quantile(.25))]
    stats={'documents':len(docs),'vocabulary_terms':len(terms),'nonzero_matrix_entries':nnz,'total_word_occurrences':int(sum(total.values())),'estimated_characters_in_tokens':int(sum(len(w)*f for w,f in total.items())),'average_word_length':round(float(mu),3),'average_document_length':round(float(df.word_count.mean()),3),'mean_term_frequency':round(float(np.mean(values)),3),'median_term_frequency':round(float(np.median(values)),3),'std_term_frequency':round(float(np.std(values)),3),'frequency_percentiles':{str(p):float(np.percentile(values,p)) for p in (25,50,75)},'vowel_consonant_ratio':round(float(vowels/consonants),4) if consonants else None,'most_common_character':chars.most_common(1),'least_common_characters':sorted([k for k,v in chars.items() if v==min(chars.values())]),'zipf_slope':round(float(slope),4),'document_length_outliers':len(outliers),'sentence_statistics_available':bool(sent_lens),'original_articles_supplied':len(textfiles),'top_terms':top[:20]}
    (OUT/'summary_statistics.json').write_text(json.dumps(stats,indent=2),encoding='utf-8')
    report=f"""# BBC News Comprehensive Text Analysis — Analysis Report

## 1. Dataset and method

The dataset contains **{len(docs):,}** BBC News documents, **{len(terms):,}** vocabulary terms and **{nnz:,}** nonzero document-term entries, organized into {len(names)} news categories. The input `.mtx` is a sparse term-frequency matrix, not raw article prose. `.terms` contains vocabulary entries (some stemmed), `.docs` contains identifiers, and `.classes` provides category labels. The analysis uses Python standard libraries, NumPy, Pandas and Matplotlib; no specialized NLP libraries.

## 2. Summary statistics

| Metric | Value |
|---|---:|
| Documents | {len(docs):,} |
| Vocabulary terms | {len(terms):,} |
| Total word occurrences | {int(sum(total.values())):,} |
| Mean words per document | {df.word_count.mean():,.2f} |
| Mean characters per token | {mu:.2f} |
| Median term frequency | {np.median(values):,.2f} |
| Standard deviation of term frequencies | {np.std(values):,.2f} |
| Vowel/consonant ratio in stored tokens | {vowels/consonants if consonants else float('nan'):.3f} |
| Zipf log-log slope | {slope:.3f} |
| Document-length upper outliers (1.5×IQR) | {len(outliers)} |

## 3. Findings and visual insights

The most frequent non-stopword terms are **{', '.join(w for w,_ in top[:10])}**. Stopwords have been removed from the top-terms charts to reveal more interpretable themes; the full unfiltered counts remain in `word_statistics.csv`. The terms may contain stemming artifacts because the vocabulary is already processed.

The rank-frequency chart (`08_zipf_law.png`) shows a strongly skewed term distribution. A fitted log-log slope of **{slope:.3f}** can be compared with the idealized Zipf exponent of -1, but this fitted value is descriptive rather than a formal goodness-of-fit test. The weighted word-length histogram shows a mean token length of **{mu:.2f}** characters; the overlaid normal curve is a visual reference, not a claim of normality.

The character-frequency charts show the relative dominance of letters and provide a vowel/consonant ratio. Punctuation may be missing because the provided term list is preprocessed. The document-length box plot identifies **{len(outliers)}** high-length outliers by the conventional 1.5×IQR rule. Category comparison and the richness scatter chart examine whether document length is associated with type-token ratio. A lower ratio in longer documents can arise mechanically from repeated words; it is not necessarily lower writing quality.

## 4. Patterns, phrases and sentiment

Email, phone, URL, hashtag, mention, date, currency and sentence regex extraction requires original character sequences. **No raw articles were included in the four input files**, so no authentic regex matches, sentence lengths or ordered n-grams can be claimed from these files. `extracted_patterns.json` records this limitation; adding actual `.txt` articles to `original_articles/` activates extraction and ordered bigrams/trigrams. The co-occurrence CSV is **document-level co-occurrence, not phrase frequency**. Keyword-based sentiment scores are exploratory counts, not validated sentiment labels.

## 5. Challenges and solutions

The main challenge is the sparse MatrixMarket format and the absence of original text. A custom streaming-compatible parser converts term/doc indices from one-based indexing and aggregates counts without constructing a huge dense matrix. The `.classes` file uses zero-based document indices; these are aligned separately. Because the dataset stores tokens rather than sentences, the implementation reports unavailable measures explicitly and supplies a distinct original-text processing path. Regex boundaries and optional phone normalization help prevent common false positives when original text is present, although real-world patterns remain more diverse than the examples.

## 6. Conclusions

The project demonstrates reliable file loading, regex functions, statistical frequency analysis, character analysis, Pandas grouping/pivots, 13 saved visualizations, Zipf analysis, document-level co-occurrence and keyword sentiment. It avoids reconstructing fictitious news prose. To complete sentence-level and regex-based evaluation on real content, add original `.txt` news articles and rerun the same script.
"""
    (ROOT/'ANALYSIS_REPORT.md').write_text(report,encoding='utf-8')
    print(json.dumps({k:v for k,v in stats.items() if k not in ('top_terms','least_common_characters')},indent=2))
    print('\nOutputs saved in:',OUT)
    print('Original raw articles:',len(textfiles),'(zero means sentence/pattern extraction unavailable)')
    print('Visualizations:',len(list(OUT.glob('*.png'))))

def interactive_dashboard():
    """Optional interactive Matplotlib category selector (desktop GUI required)."""
    df=pd.read_csv(OUT/'document_statistics.csv')
    fig,ax=plt.subplots(figsize=(10,6));plt.subplots_adjust(left=.3)
    options=['All']+sorted(df.category.unique().tolist())
    selector=RadioButtons(plt.axes([.02,.3,.22,.4]),options)
    def update(label):
        subset=df if label=='All' else df[df.category==label]
        ax.clear();ax.hist(subset.word_count,bins=30,color=COLORS[0]);ax.set_title('Document lengths — '+label);ax.set_xlabel('Words per document');ax.set_ylabel('Documents');fig.canvas.draw_idle()
    selector.on_clicked(update);update('All');plt.show()

# Run the entire analysis when this single cell executes.
analyze()
```
